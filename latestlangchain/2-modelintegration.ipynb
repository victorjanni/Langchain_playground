## Models integration with OpenAi, Google Gemini and GROQ

```python
import os 
from dotenv import load_dotenv
load_dotenv()

os.environ["OPENAI_API_KEY"]=os.getenv("OPENAI_API_KEY")
os.environ["GOOGLE_API_KEY"]=os.getenv("GOOGLE_API_KEY")
os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")
```

```python
from langchain.chat_models import init_chat_model

model = init_chat_model("gpt-4.1")
model
```

```python
response = model.invoke("what is the stack ai engineer requried?")
```

```python
response.content
```

```python
## chatopenai

from langchain_openai import ChatOpenAI

model = ChatOpenAI(model="gpt-4.1")
response= model.invoke("do you about domain specfic models?")
response.content
```

## GROQ

```python
from dotenv import load_dotenv
import os

load_dotenv(override=True)

key = os.getenv("GROQ_API_KEY")

print("Found key:", key is not None)
print("Prefix:", key[:4] if key else None)
```

```python
from langchain.chat_models import init_chat_model
model = init_chat_model("groq:qwen/qwen3-32b")
response = model.invoke("what is the process of startup hiring for ai engineers?")
response.content
```

```python
from langchain_groq import ChatGroq
model = ChatGroq(model="qwen/qwen3-32b")
response=model.invoke(" is dsa required for ai interviews")
response.content
```
